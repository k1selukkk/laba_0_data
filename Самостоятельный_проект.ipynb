Объявим функцию find_genre() для поиска неявных дубликатов в столбце с жанрами. Например, когда название одного и того же жанра написано разными словами.


```python

import pandas as pd

df = pd.read_csv('yandex_music.csv')

df.rename(columns=lambda x: x.strip().lower().replace(' ', '_'), inplace=True)
missing_values = df.isnull().sum()

df['track'] = df['track'].fillna('unknown')
df['artist'] = df['artist'].fillna('unknown')

df = df.dropna(subset=['genre'])

genres_list = df['genre'].unique()

print("Уникальные значения в столбце 'genre':")
print(genres_list)

def find_genre(genre_to_find, genre_list):

    count = 0
    for genre in genre_list:
        if genre == genre_to_find:
            count += 1
    return count

example_genre = "russian"
print(f"\nПохожие жанры на '{example_genre}':")
print(find_genre(example_genre, genres_list))
```

Создаём функцию number_tracks(), которая принимает как параметры таблицу, день недели и название города, а возвращает количество прослушанных композиций, для которых известен жанр. Проверяем количество прослушанных композиций для каждого города и понедельника, затем среды и пятницы.


```python
import pandas as pd

df = pd.read_csv('yandex_music.csv')

df.rename(columns=lambda x: x.strip().lower().replace(' ', '_'), inplace=True)
missing_values = df.isnull().sum()

df['track'] = df['track'].fillna('unknown')
df['artist'] = df['artist'].fillna('unknown')

df = df.dropna(subset=['genre'])

genres_list = df['genre'].unique()



def find_genre(genre_to_find, genre_list):

    count = 0
    for genre in genre_list:
        if genre == genre_to_find:
            count += 1
    return count



def find_hip_hop(df, incorrect_name):

    df['genre'] = df['genre'].replace(incorrect_name, 'hiphop')


    remaining_count = df['genre'].str.count(incorrect_name).sum()
    return remaining_count


incorrect_variant = 'hip-hop'
remaining_count = find_hip_hop(df, incorrect_variant)
grouped_data = df[df['genre'].notna()].groupby(['city', 'day'])['genre'].count()


def number_tracks(df, day, city):

    track_list = df[(df['day'] == day) & (df['city'] == city)]

    track_list_count = track_list['genre'].count()

    return track_list_count

print("Количество прослушанных композиций в понедельник:")
print(f"Москва: {number_tracks(df, 'Monday', 'Moscow')}")
print(f"Санкт-Петербург: {number_tracks(df, 'Monday', 'Saint-Petersburg')}")

print("\nКоличество прослушанных композиций в среду:")
print(f"Москва: {number_tracks(df, 'Wednesday', 'Moscow')}")
print(f"Санкт-Петербург: {number_tracks(df, 'Wednesday', 'Saint-Petersburg')}")

print("\nКоличество прослушанных композиций в пятницу:")
print(f"Москва: {number_tracks(df, 'Friday', 'Moscow')}")
print(f"Санкт-Петербург: {number_tracks(df, 'Friday', 'Saint-Petersburg')}")

```
